# Image retrieval and patch localization
Default: exhaustive SIFT/RANSAC. Optional DINOv2 + multi-rotation FAISS shortlisting is available through Localizer(dino=True).

```python
from localization import Localizer, make_query
import numpy as np
import cv2
from PIL import Image
from IPython.display import display
engine = Localizer(dino=False)
print(len(engine.paths), "images")
```

```python
rng = np.random.default_rng(42)
idx = int(rng.integers(len(engine.paths)))
query, truth = make_query(engine.images[idx], rng)
result = engine.match(query)
print("Known source:", engine.paths[idx].name)
print("Transform:", truth)
print("Retrieved:", result.get("image"))
print("Correct source:", result.get("image") == engine.paths[idx].name)
```

```python
display(Image.fromarray(cv2.cvtColor(query, cv2.COLOR_BGR2RGB)))
if result["accepted"]:
    print("Source image coordinates:", result["coordinates"])
    for key in ["retrieved", "localized", "aligned", "matches_image"]:
        print(key)
        display(Image.fromarray(cv2.cvtColor(result[key], cv2.COLOR_BGR2RGB)))
```
